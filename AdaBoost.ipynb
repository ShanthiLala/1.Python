```python
#The required libraries are imported at the beginning of the program.
#Pandas is used for handling datasets.
import pandas as pd
```

```python
#Load the dataset is loaded into the dataset variable using a CSV file.
dataset = pd.read_csv("insurance_pre.csv")
#print dataset and Understand the Dataset
dataset
```

```python
# Convert categorical columns to numeric
dataset = pd.get_dummies(dataset, drop_first=True)
#print dataset  
dataset
```

```python
#print dataset columns
dataset.columns
```

```python
#This step prepares the data for training.
# Assign input to independant Variable
independant = dataset[["age", "bmi", "children",  "sex_male","smoker_yes"]]
#print independant variable
independant
```

```python
# Assign output to dependendant Variable
dependendant =dataset[["charges"]]
#print dependendant variable
dependendant
```

```python
#The data is divided into: Training data and Testing data
#The training data is used to build the model and testing data is used to evaluate model performance.
#To do this we are using sklearn. sklearn.model_selection imports the train_test_split function so you can use it to split your data into training and testing sets.
from sklearn.model_selection import train_test_split
#print train_test_split
train_test_split
```

```python
# The below line of code splits data into 4 parts:
# x_train = input features for training , x_test = input features for testing
# y_train = target/output for training ,  y_test = target/output for testing
# And the options mean: test_size=0.30 → 30% of the data goes to test, 70% to train
# random_state=0 → keeps the split same every time you run it (for repeatable results)
#So Split independant and dependendant into train/test sets (70/30) in a repeatable way.   
x_train,x_test,y_train,y_test=train_test_split(independant, dependendant, test_size=0.30, random_state=0)
```

```python
y_train
Y = y_train.values.ravel()
Y
```

```python
from sklearn.ensemble import AdaBoostRegressor
from sklearn.tree import DecisionTreeRegressor
#If you do not pass estimator, it uses a decision tree regressor by default.
regressor = AdaBoostRegressor(
   # estimator=DecisionTreeRegressor(max_depth=3),
    n_estimators=100,
    learning_rate=0.1,
   # loss='linear',
    random_state=42
)

#sklearn is telling you that you passed a single column as a 2D column vector instead of a 1D target arra
# Tofix this issue  use ravel() - Flattens that array into 1 dimension.
Y = y_train.values.ravel()

regressor.fit(x_train, Y)
```

```python
# Uses the trained model to predict output values for the test inputs
y_pred = regressor.predict(x_test)
# Prints those predicted values
#y_pred
```

```python
# Imports the R² score function.
from sklearn.metrics import r2_score
# Calculates the R² value comparing actual (y_test) vs predicted (y_pred).
score = r2_score(y_test, y_pred)
# Print r2_score
score
```
